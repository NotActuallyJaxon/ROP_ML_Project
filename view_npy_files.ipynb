# Inspect Saved `.npy` Prediction Arrays & DataFrames
Loads the saved `.npy` prediction arrays alongside `occ_df` and `sev_df` to inspect shapes, summary statistics, and sample rows.

```python
from pathlib import Path
import numpy as np
import pandas as pd

# Automatically finds all .npy files in the current folder
npy_files = sorted(Path(".").glob("*.npy"))
print(f"Found {len(npy_files)} .npy file(s): {[f.name for f in npy_files]}")

arrays = {f.stem: np.load(f) for f in npy_files}
```

```python
for name, arr in arrays.items():
    print("=" * 60)
    print(f"File: {name}.npy | Shape: {arr.shape} | Dtype: {arr.dtype}")
    print(f"Min: {arr.min():.4f} | Max: {arr.max():.4f} | NaNs: {np.isnan(arr).sum()}")
    
    if arr.ndim == 2:
        df_preview = pd.DataFrame(arr, columns=[f"class_{i}" for i in range(arr.shape[1])])
        df_preview["pred_class"] = arr.argmax(axis=1)
        display(df_preview.head(-10))
    else:
        print(arr[:10])
```

```python
# Load and display the head of occ_df and sev_df
for df_name in ["occ_df", "sev_df"]:
    csv_path = Path(f"{df_name}.csv")
    parquet_path = Path(f"{df_name}.parquet")
    pkl_path = Path(f"{df_name}.pkl")
    
    if csv_path.exists():
        df = pd.read_csv(csv_path, index_col=0)
    elif parquet_path.exists():
        df = pd.read_parquet(parquet_path)
    elif pkl_path.exists():
        df = pd.read_pickle(pkl_path)
    else:
        print(f"{df_name} not found in current folder (checked .csv, .parquet, .pkl)")
        continue
        
    print("=" * 60)
    print(f"{df_name} | Shape: {df.shape}")
    display(df.head())
```
